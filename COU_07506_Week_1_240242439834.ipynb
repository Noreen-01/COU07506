```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile

plt.rcParams["figure.figsize"] = (9, 3.5)

print("DSP Laboratory is ready!")
```

```python
n = np.arange(-5, 11)

delta = (n == 0).astype(float)

step = (n >= 0).astype(float)

ramp = n * step

expo = 0.8**n * step
```

```python
fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

for ax, x, title in zip(
    axes,
    [delta, step, ramp, expo],
    ["delta(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]
):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")

plt.tight_layout()
plt.show()
```

```python
n = np.arange(-5, 11)

a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(12, 3))

for ax, a in zip(axes, a_values):
    x = (a**n) * (n >= 0)

    ax.stem(n, x)
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")

plt.tight_layout()
plt.show()
```

```python
n = np.arange(0, 41)

x1 = np.cos(2*np.pi*n/8)
x2 = np.cos(2*np.pi*n/7.5)

plt.figure(figsize=(10, 4))

plt.stem(n, x1, linefmt="b-", markerfmt="bo",
         basefmt=" ", label="cos(2πn/8)")

plt.stem(n, x2, linefmt="r-", markerfmt="ro",
         basefmt=" ", label="cos(2πn/7.5)")

plt.xlabel("n")
plt.ylabel("Amplitude")
plt.title("Comparison of Two Cosine Signals")
plt.legend()
plt.grid()
plt.show()
```

### Question 2

Why is a stem plot more appropriate than a continuous line for x(n)?

**Answer:**

A stem plot is more appropriate because a discrete-time signal is defined only at separate integer values of n. It shows each individual sample clearly. A continuous line may give the impression that the signal exists at every point in time.

### Question 3
